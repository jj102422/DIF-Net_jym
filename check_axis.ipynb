```python
import h5py
import matplotlib.pyplot as plt
```

```python
file_path = '/home/public/CTSpine1K/data/ct512/1.3.6.1.4.1.9328.50.4.0001/ct_xray_data.h5'
with h5py.File(file_path, 'r') as f:
    image = f['ct'][:]
print("image.shape:", image.shape)
```

```python
plt.figure(figsize=(8,4))
```

```python
# 获取第200层切片
# 形状应该是 (Dim1, Dim2)
slice_data = image[200, :, :]

plt.figure(figsize=(10, 8))

# --- 图 1: 完整切片 (最直观判断) ---
plt.subplot(1, 2, 1)
plt.title(f'Slice Z=200\nShape: {slice_data.shape}')
plt.imshow(slice_data, cmap='gray')
# imshow 的规则：数组的行 (Dim1) 是垂直轴，列 (Dim2) 是水平轴
plt.ylabel('Dimension 1 (Rows)\n通常对应 CT 的 Y 轴 (前-后)', fontsize=12, color='red')
plt.xlabel('Dimension 2 (Columns)\n通常对应 CT 的 X 轴 (左-右)', fontsize=12, color='blue')
plt.colorbar(fraction=0.046, pad=0.04)

# --- 图 2: 验证你提到的 image[200, :256] ---
# 如果这一步图显示图像在"垂直方向"只剩一半，说明 Dim1 确实控制垂直高度
plt.subplot(1, 2, 2)
# 这里特意取 slice_data[:256, :]，即只取 Dim1 的前256行
crop_data = image[200, :256, :] 
plt.title(f'Cropped: image[200, :256, :]\nShape: {crop_data.shape}')
plt.imshow(crop_data, cmap='gray')
plt.ylabel('Dimension 1 (0-256)', fontsize=12, color='red')
plt.xlabel('Dimension 2 (Full)', fontsize=12, color='blue')

plt.tight_layout()
plt.show()
```
